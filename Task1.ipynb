## Data Cleaning

```python
df["ReferralSource"] = df["ReferralSource"].replace({"Referral": "Other"})
df["PaymentMethod"] = df["PaymentMethod"].replace({"Online": "Other"})
```

```python
df["ReferralSource"].value_counts()
```

```python
df["PaymentMethod"].value_counts()
```

```python
df["Product"] = df["Product"].str.strip().str.title()
df["Product"].value_counts()
```

```python
df["ShippingAddress"] = df["ShippingAddress"].str.strip().str.title()
df["ShippingAddress"].value_counts()
```

```python
df["ReferralSource"] = df["ReferralSource"].str.strip().str.title()
df["ReferralSource"].value_counts()
```

### Round numeric columns to 2 decimals

```python
df["UnitPrice"] = df["UnitPrice"].round(2)
df["TotalPrice"] = df["TotalPrice"].round(2)
```

### Verify Date Format

```python
print(df["Date"].dt.strftime("%Y-%m-%d").head())
```

## Cleaning Actions Performed
The following preprocessing and cleaning operations were applied:
